```python
import gradio as gr
from gtts import gTTS
import os

def text_to_speech(text, language='en'):
    try:
        tts = gTTS(text=text, lang=language, slow=False)
        filename = "output.mp3"  # Use a fixed filename
        tts.save(filename)
        return filename
    except Exception as e:
        return f"Error: {e}"

iface = gr.Interface(
    fn=text_to_speech,
    inputs=[
        gr.Textbox(lines=2, label="Enter Text"),
        gr.Dropdown(choices=["en", "es", "fr", "de", "ja", "ko", "pt", "ru", "zh-CN", "hi"], value="en", label="Language") #Added language selection
    ],
    outputs=gr.Audio(label="Generated Audio"),
    title="Text-to-Speech with gTTS",
    description="Convert text to speech using gTTS. Select the language from the dropdown menu.",
    examples=[
        ["Hello, this is a test.", "en"],
        ["Hola, esto es una prueba.", "es"],
        ["Bonjour, ceci est un test.", "fr"]
    ]
)

iface.launch(debug=True, share=True)
```
